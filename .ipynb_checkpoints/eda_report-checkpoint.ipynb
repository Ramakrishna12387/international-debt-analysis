```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Cleaned data load
df = pd.read_csv("cleaned_international_debt.csv")

# 1. Summary Statistics
print("Dataset Overview:")
print(df.info())
print("\nStatistical Summary:")
print(df.describe())

# 2. Top 10 Countries by Total Debt (Matplotlib/Seaborn)
plt.figure(figsize=(10, 5))
top_countries = df.groupby('country_name')['debt'].sum().sort_values(ascending=False).head(10)
sns.barplot(x=top_countries.values, y=top_countries.index, palette='magma')
plt.title('Top 10 Countries by Total External Debt')
plt.xlabel('Total Debt (USD)')
plt.ylabel('Country')
plt.tight_layout()
plt.show()

# 3. Debt Distribution Across Indicators
plt.figure(figsize=(10, 5))
top_indicators = df.groupby('indicator_name')['debt'].sum().sort_values(ascending=False).head(5)
sns.barplot(x=top_indicators.values, y=top_indicators.index, palette='viridis')
plt.title('Top 5 Debt Indicators Contributing to Global Debt')
plt.xlabel('Debt Amount (USD)')
plt.ylabel('Indicator')
plt.tight_layout()
plt.show()
```
